```python
"""Defines the variables to avoid an error."""
total_amount = 0

amount_consumed = ""


"""Definition of function for converting units"""
def convert_to_ml(amount, unit):

  if unit == "l":
    return amount * 1000
    # Converts amount to ml by multiplying 1000

  elif unit == "ml":
    return amount


"""Uses a loop to repeatedly ask for the amount consumed.
Allows the user to enter 0 to stop recording."""
while True:

  try:

    amount_consumed = int(input("Enter amount consumed (0 to stop): "))


    if amount_consumed < 0:
      print("Please enter a valid input!\n")
    # Rejects negative inputs.

    else:

      if amount_consumed == 0:
        break
      # Stops recording when 0 is entered.

      else:
        unit = input("Unit (ml/l): ")
        # Asks user for the unit.

        try:

          total_amount = total_amount + convert_to_ml(amount_consumed, unit)
          # Adds each returned value to a running total.

          print("")
          # Line break to separate between inputs of values.

        except TypeError or ValueError:

          print("Please enter a valid input!\n")
        # Asks the user to try again if unit is invalid.

  except ValueError:

      print("Please enter a valid input!\n")
      # Asks the user to try again if number is invalid.


"""Prints the total water intake in both ml and liters."""
print(f"\nTotal water intake: {total_amount} ml")
print(f"Equivalent: {total_amount/1000} liters")
```
