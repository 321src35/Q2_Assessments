```python
print("Enter the 2 numbers you want to operate with:")
# Asks the user to input 2 numbers for operation

number_prompt_loop = True

while number_prompt_loop:
  try:
    first_number = int(input("First number: "))
    second_number = int(input("Second number: "))
    number_prompt_loop = False
  except ValueError:
    print("Please input a number!")
# Prompts the user for each number, asks again if invalid

def add(num1, num2):
  return (num1 + num2)
def subtract(num1, num2):
  return (num1 - num2)
def multiply(num1, num2):
  return (num1 * num2)
def divide(num1, num2):
  return (num1 / num2)
# Function definition for each operation

print("Choose your operation:")
print("1 = Addition")
print("2 = Subtraction")
print("3 = Multiplication")
print("4 = Division")
# Tells the user to choose an operation by typing a number from 1-4

operation_prompt_loop = True
while operation_prompt_loop:
  try:
    opr = int(input("Operation (1-4): "))
    if 1 <= opr <= 4:
      operation_prompt_loop = False
    else:
      print("Invalid operation!")
  except ValueError:
    print("Please input an integer from 1-4!")
# Prompts the user to choose an operation, asks again if invalid

if opr == 1:
  print("You have chosen: Addition.")
elif opr == 2:
  print("You have chosen: Subtraction.")
elif opr == 3:
  print("You have chosen: Multiplication.")
elif opr == 4:
  print("You have chosen: Division.")
# Tells the user his/her chosen operation

if opr == 1:
  print(f"{first_number} + {second_number} is:")
  add(first_number, second_number)
elif opr == 2:
  print(f"{first_number} - {second_number} is:")
  subtract(first_number, second_number)
elif opr == 3:
  print(f"{first_number} * {second_number} is:")
  multiply(first_number, second_number)
elif opr == 4:
  try:
    print(f"{first_number} / {second_number} is:")
    divide(first_number, second_number)
  except ZeroDivisionError:
    print("Cannot divide by zero!")
# Prints the user's given numbers and final answer in an equation
```

## 1. What functions did you create in your calculator program?
I created functions for addition, subtraction, multiplication, and division between two numbers.

## 2. What parameters did your functions use?
My functions used num1 and num2 as parameters.

## 3. What arguments were passed when the functions were called?
The arguments first_number and second_number were passed when the functions were called.

## 4. How did your program use the returned value?
My program used the returned value by displaying it when its function was called.

## 5. Why is it better to divide the calculator program into functions instead of writing everything in one long block of code?
It's better to have functions instead of one long block of code because using functions in your code makes it much more organized and neat. It also makes the code much easier to understand and modify.
